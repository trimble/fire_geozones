```python
# Cell 2: Imports
import geopandas as gpd
import osmnx as ox
import matplotlib.pyplot as plt
import numpy as np
from shapely.geometry import Point, Polygon
from shapely.ops import unary_union
from itertools import combinations
```

```python
# Cell 3: Load township boundary (replace with your file path)
townships = gpd.read_file(r'~/workspace/fire_voronoi/Local_Boundaries_Minor_Civil_Divisions/Minor_Civil_Divisions_IN.shp')

# Reproject to WGS84 (EPSG:4326), required for osmnx
if townships.crs is None:
    townships.set_crs(epsg=4326, inplace=True)
elif townships.crs.to_epsg() != 4326:
    townships = townships.to_crs(epsg=4326)
is_shelby = townships['Cnty_Name']=="Shelby"
shelby_townships = townships[is_shelby]
township_union = shelby_townships.unary_union
bounds = shelby_townships.total_bounds  # [minx, miny, maxx, maxy]
```

```python
# Cell 4: Load road data from OSM using township polygon
import osmnx as ox

# Ensure your township is a single polygon
G = ox.graph_from_polygon(township_union, network_type='drive')
roads = ox.graph_to_gdfs(G, nodes=False, edges=True)
```

```python
# Cell 5: Define fire station coordinates (EDIT HERE)
# Format: [(lon, lat), ...]
fire_station_coords = [
  (-85.8572277, 39.5844416),
  (-85.9070139, 39.6387254),
  (-85.9709754, 39.529953),
  (-85.8810707, 39.4406702),
  (-85.7786288, 39.5227097),
  (-86.0065276, 39.6691316),
  (-85.9990409, 39.6040567),
  (-85.779439, 39.6937218),
  (-85.9795233, 39.6962247),
  (-85.7045637, 39.6760901),
  (-85.6662435, 39.4537326),
  (-85.8309059, 39.3641328),
  (-85.9131422, 39.7338448),
  (-85.617652, 39.5710743),
  (-86.0747433, 39.5484272),
  (-85.7434899, 39.51627),
  (-85.7945173, 39.5055857),
  (-86.0737878, 39.523629),
  (-86.1040209, 39.6134433),
  (-85.6285885, 39.4282778),
  (-85.896068, 39.7762501),
  (-86.06305093068505, 39.625805791286666),
  (-86.09102800041359, 39.5577404582559),
  (-86.02353331575846, 39.482230271604614)
]
```

```python
fire_stations = gpd.GeoDataFrame(
    geometry=[Point(xy) for xy in fire_station_coords],
    crs="EPSG:4326"
).to_crs(epsg=3857)
```

```python
# Cell 6: Compute second-order Voronoi diagram (approximate method)
def second_order_voronoi(points, boundary_polygon):
    coords = np.array([pt.coords[0] for pt in points])
    pairs = list(combinations(range(len(points)), 2))
    result = {}

    for i, j in pairs:
        pi, pj = coords[i], coords[j]
        midpoint = (pi + pj) / 2
        region = Point(midpoint).buffer(2500)  # Buffer of ~2.5 km
        clipped = region.intersection(boundary_polygon)
        result[(i, j)] = clipped

    return result

second_order_regions = second_order_voronoi(fire_stations.geometry, township_union)
```

```python
# Cell 7: Plotting
fig, ax = plt.subplots(figsize=(10, 10))

townships.boundary.plot(ax=ax, color='black', linewidth=1, label='Township')
roads.plot(ax=ax, color='gray', linewidth=0.5, alpha=0.7, label='Roads')
fire_stations.plot(ax=ax, color='red', markersize=50, label='Fire Stations')

# Plot second-order Voronoi regions
for (i, j), region in second_order_regions.items():
    if region.is_empty:
        continue
    if region.geom_type == 'Polygon':
        ax.fill(*region.exterior.xy, alpha=0.3, label=f'FS{i}-FS{j}')
    elif region.geom_type == 'MultiPolygon':
        for poly in region.geoms:
            ax.fill(*poly.exterior.xy, alpha=0.3, label=f'FS{i}-FS{j}')

# Label fire stations
for idx, row in fire_stations.iterrows():
    ax.annotate(f'FS{idx}', (row.geometry.x + 50, row.geometry.y + 50), fontsize=10)

ax.set_title("Second-Order Voronoi Regions by Fire Station Pairs")
ax.set_axis_off()
plt.legend()
plt.show()
```
